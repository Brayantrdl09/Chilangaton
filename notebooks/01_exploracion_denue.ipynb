```python
from pathlib import Path
import zipfile

ruta = Path("../Datos/originales/denue")

print("¿Existe la carpeta?:", ruta.exists())

archivos = list(ruta.glob("*"))

for archivo in archivos:
    print(archivo.name)

    if archivo.suffix.lower() == ".zip":
        with zipfile.ZipFile(archivo) as zip:
            for nombre in zip.namelist():
                print("  ", nombre)
```

```python
import pandas as pd
import zipfile

ruta = "../Datos/originales/denue/denue_09_csv.zip"

with zipfile.ZipFile(ruta) as zip:
    with zip.open("conjunto_de_datos/denue_inegi_09_.csv") as archivo:
        df = pd.read_csv(archivo, encoding="latin-1", low_memory=False)

print("Filas:", df.shape[0])
print("Columnas:", df.shape[1])

display(df.head())
```

```python
print(df.columns.tolist())
```

```python
# Filtrar comercio al por menor de abarrotes y alimentos
df_4611 = df[
    df["codigo_act"].astype(str).str.startswith("4611")
].copy()

print("Total de establecimientos:", len(df_4611))

display(df_4611.head())
```

```python
df_4611.groupby(
    ["codigo_act", "nombre_act"]
).size().reset_index(name="Establecimientos")
```

```python
from pathlib import Path

ruta_salida = Path("../Datos/procesados")
ruta_salida.mkdir(parents=True, exist_ok=True)

# Guardar en Parquet
df_4611.to_parquet(
    ruta_salida / "denue_4611_cdmx_2026.parquet",
    index=False
)

# Guardar en CSV
df_4611.to_csv(
    ruta_salida / "denue_4611_cdmx_2026.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Archivos guardados correctamente")
print("Registros:", len(df_4611))
print("Columnas:", len(df_4611.columns))
```
