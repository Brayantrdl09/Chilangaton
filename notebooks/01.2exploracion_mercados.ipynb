```python
import geopandas as gpd

ruta = "../Datos/originales/mercados/mercados_cdmx_sedeco/MERCADOS_ADIP.shp"

gdf_mercados = gpd.read_file(ruta)

print("Registros:", len(gdf_mercados))
print("Columnas:", gdf_mercados.columns.tolist())
print("Sistema de coordenadas:", gdf_mercados.crs)

display(gdf_mercados.head())
```

```python
# Valores faltantes
print("VALORES FALTANTES")
print(gdf_mercados.isna().sum())

# Duplicados exactos
print("\nDUPLICADOS EXACTOS")
print(gdf_mercados.duplicated().sum())

# Nombres de mercados repetidos
print("\nNOMBRES REPETIDOS")
print(gdf_mercados["MERCADO"].value_counts().head(15))

# Mercados por alcaldía
print("\nMERCADOS POR ALCALDÍA")
print(gdf_mercados["DELEGACION"].value_counts())

# Tipos de mercado
print("\nTIPOS")
print(gdf_mercados["TIPO"].value_counts(dropna=False))
```

```python
print("COORDENADAS FALTANTES")
print(gdf_mercados[["COORGEOG_X", "COORGEOG_Y"]].isna().sum())

print("\nRANGOS GEOGRÁFICOS")
print(gdf_mercados[["COORGEOG_X", "COORGEOG_Y"]].describe())

print("\nGEOMETRÍAS VACÍAS")
print(gdf_mercados.geometry.is_empty.sum())

print("\nGEOMETRÍAS NULAS")
print(gdf_mercados.geometry.isna().sum())

print("\nTIPOS DE GEOMETRÍA")
print(gdf_mercados.geometry.geom_type.value_counts())
```

```python
# Revisar identificadores repetidos
print("NUM_OFIC duplicados:")
print(gdf_mercados["NUM_OFIC"].duplicated().sum())

# Revisar posibles ubicaciones repetidas
print("\nCoordenadas repetidas:")
print(
    gdf_mercados.duplicated(
        subset=["COORGEOG_X", "COORGEOG_Y"]
    ).sum()
)

# Comparar coordenadas con geometría
diferencia_x = (
    gdf_mercados.geometry.x -
    gdf_mercados["COORGEOG_X"]
).abs()

diferencia_y = (
    gdf_mercados.geometry.y -
    gdf_mercados["COORGEOG_Y"]
).abs()

print("\nDiferencia máxima en longitud:", diferencia_x.max())
print("Diferencia máxima en latitud:", diferencia_y.max())

# Revisar valores N/D
print("\nValores N/D por columna:")
print(gdf_mercados.astype(str).eq("N/D").sum())
```

```python
# Copia para realizar las verificaciones
revision = gdf_mercados.copy()

# Diferencias entre geometría y coordenadas
revision["dif_lon"] = (
    revision.geometry.x - revision["COORGEOG_X"]
).abs()

revision["dif_lat"] = (
    revision.geometry.y - revision["COORGEOG_Y"]
).abs()

# Mercados con diferencias importantes
problemas_geo = revision[
    (revision["dif_lon"] > 0.001) |
    (revision["dif_lat"] > 0.001)
]

print("MERCADOS CON DIFERENCIAS GEOGRÁFICAS")
print("Total:", len(problemas_geo))

display(
    problemas_geo[
        ["DELEGACION", "MERCADO", "DIRECCION",
         "COORGEOG_X", "COORGEOG_Y", "dif_lon", "dif_lat"]
    ]
)

# Identificadores oficiales repetidos
print("\nIDENTIFICADORES REPETIDOS")

display(
    revision[
        revision.duplicated("NUM_OFIC", keep=False)
    ][["DELEGACION", "MERCADO", "NUM_OFIC", "DIRECCION"]]
)

# Coordenadas repetidas
print("\nCOORDENADAS REPETIDAS")

display(
    revision[
        revision.duplicated(
            ["COORGEOG_X", "COORGEOG_Y"],
            keep=False
        )
    ][["DELEGACION", "MERCADO", "DIRECCION",
       "COORGEOG_X", "COORGEOG_Y"]]
)
```

```python
# Registros que presentan inconsistencias
indices = [23, 54, 266, 284]

comparacion = gdf_mercados.loc[indices].copy()

# Coordenadas provenientes de la geometría
comparacion["lon_geometry"] = comparacion.geometry.x
comparacion["lat_geometry"] = comparacion.geometry.y

columnas = [
    "DELEGACION",
    "MERCADO",
    "NUM_OFIC",
    "DIRECCION",
    "COLONIA",
    "COORGEOG_X",
    "COORGEOG_Y",
    "lon_geometry",
    "lat_geometry"
]

display(comparacion[columnas])
```

```python
import pandas as pd

# Copiar la base original
gdf_mercados_final = gdf_mercados.copy()

# Guardar coordenadas originales de los atributos
gdf_mercados_final["longitud_original"] = gdf_mercados_final["COORGEOG_X"]
gdf_mercados_final["latitud_original"] = gdf_mercados_final["COORGEOG_Y"]

# Extraer coordenadas de la geometría
gdf_mercados_final["longitud"] = gdf_mercados_final.geometry.x
gdf_mercados_final["latitud"] = gdf_mercados_final.geometry.y

# Identificar discrepancias
gdf_mercados_final["revision_pendiente"] = (
    (gdf_mercados_final["longitud"] -
     gdf_mercados_final["longitud_original"]).abs() > 0.001
) | (
    (gdf_mercados_final["latitud"] -
     gdf_mercados_final["latitud_original"]).abs() > 0.001
)

# Convertir N/D en valores nulos
gdf_mercados_final = gdf_mercados_final.replace("N/D", pd.NA)

print("Mercados:", len(gdf_mercados_final))
print("Pendientes de revisión:", gdf_mercados_final["revision_pendiente"].sum())
```

```python
from pathlib import Path

ruta_salida = Path("../Datos/procesados")
ruta_salida.mkdir(parents=True, exist_ok=True)

# CSV sin la columna geométrica
gdf_mercados_final.drop(columns="geometry").to_csv(
    ruta_salida / "mercados_cdmx_sedeco.csv",
    index=False,
    encoding="utf-8-sig"
)

# GeoParquet con la geometría
gdf_mercados_final.to_parquet(
    ruta_salida / "mercados_cdmx_sedeco.parquet",
    index=False
)

print("Archivos guardados correctamente")
```
